```python
# pie chart - top 3 countries that uses zomato
plt.pie(country_values[:3],labels=country_names[:3],autopct='%1.1f%%')
```

```python
# same with better visuals
# Adjusting figure size
plt.figure(figsize=(10, 6))

# Define an 'explode' to separate the India slice
my_explode = [0.1, 0, 0] 

plt.pie(country_values[:3], 
        labels=country_names[:3], 
        autopct='%1.2f%%',    # This adds the percentage
        pctdistance=0.85,     # Moves the % numbers further from the center
        explode=my_explode, 
        shadow=True, 
        colors=['#ff9999','#66b3ff','#99ff99'],
        startangle=140,
        textprops={'fontsize': 12, 'weight': 'bold'}) # Makes text easier to read

plt.title('Top 3 Countries Using Zomato', fontsize=15, pad=20)
plt.axis('equal') 
plt.show()
```

```python
rating=final.groupby(['Aggregate rating','Rating color','Rating text']).size().reset_index().rename(columns={0:'Rating count'})
```

```python
rating
```

## observations
1. when rating is between 4.5 to 4.9 ---> Excellent
2. when rating is between 4 to 4.4 ---> Very Good
3. when rating is between 3.5 to 3.9 ---> Good
4. when rating is between 2.5 to 3.4 ---> Average
5. when rating is between 1.8 to 2.4 ---> Poor

```python
# bargraph
import matplotlib
matplotlib.rcParams['figure.figsize']=(12,6)
sns.barplot(x='Aggregate rating',y='Rating count',hue='Rating color',data=rating,palette=['blue','red','orange','yellow','green','green'])
```

```python
# diagram
import matplotlib.pyplot as plt
import seaborn as sns

# Set the figure size
plt.figure(figsize=(12, 6))

# Map the 'Rating color' column to actual matplotlib colors
# Note: 'white' is replaced with 'blue' for better visibility on the plot
sns.barplot(x="Aggregate rating", y="Rating count", data=rating, hue='Rating color', 
            palette=['blue', 'red', 'orange', 'yellow', 'green', 'darkgreen'], legend=False)

# Add title and labels
plt.title('Distribution of Ratings', fontsize=15)
plt.xlabel('Aggregate Rating', fontsize=12)
plt.ylabel('Rating Count', fontsize=12)

plt.show()
```

## obsevations
1. not rated count is very high
2. maximum number of rating are between 2.5 to 3.4

```python
# count plot
sns.countplot(x='Rating color',data=rating,palette=['blue','red','orange','yellow','green','green'])
```
